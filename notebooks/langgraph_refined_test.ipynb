Importación de librerias y dependencias

```python
from dotenv import load_dotenv
from langchain.messages import HumanMessage
from rafael_langgraph_websearchagent.langgraph_agent_refined import agent
load_dotenv()
```

Prompt 1: Pregunta de conocimientos estaticos

```python
prompt = "¿Quien es Doraemon?"

respuesta = agent.invoke({"messages": [HumanMessage(content=prompt)]})

#Mostrar la conversación en pantalla
print("=== FLUJO DE CONVERSACIÓN ===")
for msg in respuesta["messages"]:
    msg.pretty_print()

#Mostrar las métricas de uso de tokens
print("\n" + "=" * 40)
print("📊 MÉTRICAS DE USO DE TOKENS")
print("=" * 40)
print(f"🔹 Llamadas al LLM    : {respuesta.get('llm_calls', 0)}")
print(f"🔹 Tokens de Prompt    : {respuesta.get('prompt_tokens', 0)}")
print(f"🔹 Tokens de Respuesta : {respuesta.get('completion_tokens', 0)}")
print(f"🔹 Tokens Totales      : {respuesta.get('total_tokens', 0)}")
print(f"🛠️ Tokens en Tool Calls: {respuesta.get('tool_tokens', 0)}")
print("=" * 40)
```
