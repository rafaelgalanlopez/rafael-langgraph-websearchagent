Importar dependencias y el agente

```python
from dotenv import load_dotenv
from langchain.messages import HumanMessage
from rafael_langgraph_websearchagent.langgraph_agent import agent

# Cargar las variables de entorno del archivo .env (claves de API)
load_dotenv()
```

Prompt 1: Utilizar herramienta de busqueda

```python
# Definimos una consulta sobre eventos o datos recientes
prompt_busqueda = HumanMessage(content="¿Cuáles son los últimos avances o noticias sobre el lanzamiento de Python 3.13?")

response_1 = agent.invoke({"messages": [prompt_busqueda]})

# Imprimimos el flujo de mensajes para verificar el uso de la herramienta
for msg in response_1["messages"]:
    msg.pretty_print()
```

Prompt 2: Sin herramientas

```python
# Definimos una consulta de conocimiento general/teórico
prompt_sin_herramienta = HumanMessage(content="Explícame qué es la programación orientada a objetos en 3 frases sencillas.")

response_2 = agent.invoke({"messages": [prompt_sin_herramienta]})

# Imprimimos el flujo de mensajes
for msg in response_2["messages"]:
    msg.pretty_print()
```
